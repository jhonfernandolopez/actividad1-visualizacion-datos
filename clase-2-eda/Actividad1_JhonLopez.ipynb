## Parte 3 — Construir gráficos

Construye al menos **3 de los 5 tipos de gráfico** que vimos en la Clase 2, usando tus propias columnas. Reemplaza `columna_x`, `columna_y`, etc. por los nombres reales de tu dataset.

### Gráfico 1 — Barras (`geom_col`)

```r
# Afiliados por estamento (se agrega primero con count y se ordena)
mis_datos |>
  count(descripcion_estamento) |>
  ggplot(aes(x = n, y = reorder(descripcion_estamento, n))) +
  geom_col(fill = "steelblue") +
  labs(title = "Afiliados activos por estamento",
       x = "Número de afiliados", y = "Estamento")
```

**¿Este gráfico es de análisis o de presentación? ¿Por qué?**
De presentación: está ordenado de mayor a menor, con título y ejes con nombre propio, listo para que alguien que no conoce los datos entienda de una vez cuáles estamentos concentran la afiliación (empleados públicos docentes y no docentes, seguidos de los pensionados).

### Gráfico 2 — Dispersión (`geom_point`)

```r
ggplot(data = mis_datos, aes(x = antiguedad_anios, y = edad)) +
  geom_point(color = "firebrick", alpha = 0.15) +
  labs(title = "Edad vs. antigüedad de afiliación",
       x = "Antigüedad de afiliación (años)", y = "Edad (años)")
```

**¿Este gráfico es de análisis o de presentación? ¿Por qué?**
De análisis: lo hice para **buscar** una relación, no para comunicarla. El alpha bajo deja ver la densidad de los 5.454 puntos y aparecen patrones que no esperaba: una franja diagonal (gente afiliada casi desde el nacimiento, donde edad ≈ antigüedad) y grupos de beneficiarios jóvenes con poca antigüedad.

### Gráfico 3 — Boxplot (`geom_boxplot`)

```r
ggplot(data = mis_datos, aes(x = sexo, y = edad, fill = sexo)) +
  geom_boxplot() +
  labs(title = "Distribución de la edad por sexo",
       x = "Sexo", y = "Edad (años)")
```

**¿Este gráfico es de análisis o de presentación? ¿Por qué?**
De análisis: el boxplot compara de un vistazo mediana, cuartiles y atípicos entre los dos grupos. Las distribuciones son parecidas y con medianas altas (alrededor de los 60 años), con presencia de menores (beneficiarios) como valores bajos.

### Gráfico 4 (opcional) — Histograma (`geom_histogram`)

```r
ggplot(data = mis_datos, aes(x = edad)) +
  geom_histogram(bins = 15, fill = "darkgreen", color = "white") +
  labs(title = "Distribución de la edad de los afiliados activos",
       x = "Edad (años)", y = "Frecuencia")
```

### Gráfico 5 (opcional) — Líneas (`geom_line`)

*Solo si tu dataset tiene una variable de tiempo (fechas, meses, años).*

```r
# Afiliaciones nuevas por año (1969-2026)
mis_datos |>
  count(anio_afiliacion) |>
  ggplot(aes(x = anio_afiliacion, y = n, group = 1)) +
  geom_line(color = "firebrick") +
  geom_point() +
  labs(title = "Afiliaciones por año de vinculación",
       x = "Año de afiliación", y = "Afiliaciones")
```

---
## 🎯 Conclusión

La población afiliada activa está claramente envejecida: la mediana de edad es de 64 años, la antigüedad mediana de afiliación supera los 20 años y los estamentos de pensionados pesan casi tanto como los de empleados activos. La dispersión edad-antigüedad revela dos poblaciones conviviendo: cotizantes de décadas de vinculación y beneficiarios jóvenes recientes. Para el servicio de salud esto confirma con datos algo que se intuye en la operación diaria: el perfil de demanda es el de una población mayor.
